```python
import pandas as pd
import os

folder = r'/content/drive/My Drive/Coursera- Data Analytics Files/Wellness Company Case Study/Data/archive (2)'

file_info = []

# Use os.walk to recursively find all CSV files in the folder and its subdirectories
for root, dirs, files in os.walk(folder):
    for file in files:
        if file.endswith(".csv"):
            file_path = os.path.join(root, file)
            try:
                df = pd.read_csv(file_path)
                # Get the relative path for the 'File' column, starting from the original 'folder'
                relative_path = os.path.relpath(file_path, folder)
                file_info.append({
                    "File": relative_path,
                    "Rows": len(df),
                    "Columns": len(df.columns)
                })
            except pd.errors.EmptyDataError:
                print(f"Warning: {file_path} is empty and was skipped.")
            except Exception as e:
                print(f"Error reading {file_path}: {e}")

file_inventory = pd.DataFrame(file_info)

# Sort by 'File' for better readability
if not file_inventory.empty:
    file_inventory = file_inventory.sort_values(by="File").reset_index(drop=True)

print(file_inventory)
```
